## Nested Dictionary

#### Nested dictionary will be used for keeping a dictionary inside another dictionary

```python
## I want to store information for multiple student like their name , marks , grade
nested_dictionary={"Student_1":{"Name":"Oly","Marks":90,"Grade":"A+"}
                   , "Student_2":{"Name":"Jaya","Marks":95,"Grade":"A"},
                       "Student_3":{"Name":"Noel","Marks":80,"Grade":"A"}
                   }
```

```python
nested_dictionary
```

```python
type(nested_dictionary)
```

```python
## nested_dictionary.items()  call must be .items()
for outer_key, outer_value in nested_dictionary.items():
    print(f"Outer Key is: {outer_key} and Outer Value is: {outer_value}")

    for key_1, value_1 in outer_value.items():
        print(f"Inner Key is: {key_1} and Inner Value is: {value_1}")
```

## dictionary Comprehension 
## Dictionary helps us with using looping statement along with conditional statement inside our dictionary

## syntax for dictionary comprehension:
## {Expression for item in iterable if condiiton}

```python
comprehensive={x:x**2 for x in range(11)}
```

```python
comprehensive
```

```python
even_dictionary={x:x**2 for x in range(1,11) if(x%2)==0}
odd_dictionary={x:x**2 for x in range(1,11) if(x%2)!=0}
```

```python
print(even_dictionary)
print(odd_dictionary)
```

#### merged Dictionary

```python
dictionary_1=even_dictionary.copy()
dictionary_2=odd_dictionary.copy()
```

```python
merged_dictionary={**dictionary_1,**dictionary_2}
```

```python
merged_dictionary
```
