# Tuples

### Tuples are the second type of data structure which can store the data in a ordered way just like list but tuples are not mutable(not to option to replace value)

### tuples will be created inside these brackets ()

```python
a=tuple()
```

```python
type(a)
```

```python
tuple_values=(1,2,3,4,5)
print(tuple_values)
```

### Tuples are like list only which can store data inside it a ordered way  but are going to be immutable

```python
tuple_values[1]
```

```python
tuple_values[1]=24
```

```python
list_values=[1,2,3,4]
print(list_values)
print(list_values[2])
list_values[2]=34
print(list_values[2])

```

### Tuples are also capable of storing the data of any data type

```python
tuple_data=(1,2,3,4,5,6,"B","A","T",4.5,True)
```

```python
### tuples are going to be ordered or are going to be indexed 

tuple_data[:]
```

```python
tuple_data[1:]
```

```python
tuple_data[::-1]
```

```python
tuple_data[1:4]
```

```python
tuple_data[0:4]
```

### Methods In Tuples

## tuples we are going to have 2 methods usually say as :

## 1. count()
## 2. index() 

```python
tuple_method=(1,2,3,4,8,6,9,2,4,3,1,2,2,3,4,6,8)
```

```python
tuple_method.index(2) ### always return 1st index
```

```python
tuple_method.count(2)
```
