```python
a=tuple()
```

```python
type(a)
```

```python
tuple_values=(1,2,3,4,5)
print(tuple_values)
```

### Tuples are like list only which can store data inside it a ordered way  but are going to be immutable

```python
tuple_values[1]
```

```python
tuple_values[1]=24
```

```python
list_values=[1,2,3,4]
print(list_values)
print(list_values[2])
list_values[2]=34
print(list_values[2])

```
