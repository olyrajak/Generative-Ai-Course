## List comprehension 
### we are going to use looping statement inside the list so that we can store the value inside the list in an iterable format and in list comprehesion condition statements are allowed as well.



## syntax for List comprehension is 
### [expression for item in iterable if condition ]

### expresssion ----> what we want to print 
### for item in iterable ----> looping statement
### if condition ----> conditional statement

```python
# I need to store around 20 valueues. I need to store value from 1 to 50 
```

```python
list_items=[x for x in range(1,51)]
print(list_items)
```

```python
list_squared=[x**2 for x in range(1,16)]
print(list_squared)
```

```python
# If I only interested in lookibg over the square values which are evn 
even_items=[x for x in range(1,19) if x%2==0]
print(even_items)
```

```python
# If I only interested in lookibg over the square values which are odd 
odd_items=[x for x in range(1,19) if x%2!=0]
print(odd_items)
```
