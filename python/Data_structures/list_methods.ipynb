# List Method

## List method is the append method
### append()

```python
list_items=[1,2,4,7,8,8,9]
```

```python
### append() -----> insert value at the end of the list
list_items.append(10)
print(list_items)
```

```python
### Insert a value at specific locaion.
list_items.insert(3,24) # 3 is index and 24 is value
```

```python
print(list_items)
```

```python
# remove 
list_items.remove(8) # remove first occurence of 8
print(list_items)
```

```python
list_items.remove(8)
```

```python
print(list_items)
```

```python
# pop 
list_items.pop() # remove last element defaut
print(list_items)
```

```python
# pop 
list_items.pop(2) # remove last element defaut
print(list_items)
```
