# Operators

## Operators helps in applying operation on the given operads

## Types Of Operators :
1. Arithmetic Operators 

Addition (+)
Substraction (-)
Multiplication (*)
Division (/)
Modulus (%) ---- find remainder
Floor  (//) --- Apply the floor operation
Exponent (a**b) ----This means "a raised to the power of b". print(2 ** 3)

2. Comparison Operators 

Greater than (>)
Less than (<)
Greater than or equal to (>=)
Less than or euqal to (<=)
Equality Check (==)
Not Equality Check (!=)

3.Logical Operators

and ----> whether if both condition true or not 

or ----> whether if any of the operation is true or not

not ----> negation or to inverse the result of the given operation 




## Arithmetic Operators

```python
# Addition Operators
a=5
b=6
print("Addition of a and b is:", a + b)
```

```python
# Substraction operators
a=15
b=10
print("Substraction of a and b is:", a - b)
```

```python
# Product Or Multiplication operators
a=10
b=5
print("Product of a and b is:", a * b)
```

```python
#Division operators
a=10
b=2
print("Division of a and b is:", a / b)
```

```python
#Floor operators remove decimal part and return the integer part of the result
a=18
b=5
print(a//b)
```

```python
# Modulus Operators
## Modulus Operators will be helping us out with getting the idea about the remainder that will be 
# left after dividing the first value with the second value
a=30
b=7
print("Modulus of a and b is:", a % b)
```

```python
#Exponential Operators
## Exponential Operators will be helping us out with getting the idea about the power of the first value raised to the second value

a=2
b=3
print("Exponential of a and b is:", a ** b)
```

# Comparison Operators
## Comparison Operators are used to compare two values and return a boolean value (True or False) based on the comparison result


```python
# Euality Check
a=10
b=20
a==b
print("Is a equal to b?", a == b)
print(a)
```

```python
# Not Equal Check
a=10
b=20
a==b
a!=b
print("Is a not equal to b?", a != b)
```

```python
# Greater than Check
a=3
b=5
print("Is a greater than b?", a > b)
```

```python
# Less than Check
a=5
b=2
print("Is a less than b?", a < b)
```

```python
#greater than or equal to Check
a=5
b=5
print("Is a greater than or equal to b?", a >= b)
a=10
b=20
print("Is a greater than or equal to b?", a >= b)

```

```python
# Less than or equal to Check
a=5
b=5
print("Is a less than or equal to b?", a <= b)
a=4
b=1
print("Is a less than or equal to b?", a <= b)
```

# Logical Operators

```python
#and operators

a=10
b=6
c=7

print(a>b and a>c)
print(a >10 and b>c)
```

```python
#or operators
print(a>b or a>c)
print(a >b or b>c) 
```

```python
# not operators
a=True
print(not(a))
```

```python
# Simple Calculator
a=int(input("Enter the first number: "))
b=int(input("Enter the second number: "))
operation=input("Enter the operation you want to perform (+, -, *, /): ")
if operation == '+':
    print("The result of addition is:", a + b)
elif operation == '-':
    print("The result of subtraction is:", a - b)
elif operation == '*':
    print("The result of multiplication is:", a * b)
elif operation == '/':
    print("The result of division is:", a / b)
else:
    print("Invalid operation")
```
