```python
## by making the use of while loop we need to print the number 0 to 50
i=0
while i<50:
    print(i)
    i=i+1
```

# Loop Control Statement

## Loop control statement are being used for terminating a loop in between when a condition is meet or if you want to have some control over your iterating  process while iterating over the series of numbers

### Break ----> break is our first loop control statement which will help you out to terminate the loop when a certain condition is meet  

```python
## I want to print the numbers from 0 to 50 but once number is 18 then terminate the loop or do npt print 
## either 18 or any number forward 


for i in range(51):
    if i==18:
        break
    else :
        print(i)
```

### Continue ---> continue is our second loop control statement which helps us in iterating over a series of numbers but helps us with skipping the values in between once we meet a condition


```python
## print the number from 0 to 50 but only print even number and for odd numbers skip those values 

for i in range(51):
    if i%2==0:
        print(i)
    else:
        continue
```

```python
## Print the number from 0 to 20 but once 17 then skip it 

for i in range(21):
    if i==17:
        continue
    else:
         print(i)
```

### Pass --->statement is being used for doing nothing 

```python
## print the numbers from 0 to 20 but once 17 is occuring do nothing 

for i in range(21):
    if i==17:
        pass
        print("Our programme will do nothing since the value is 17")

    else:
        print(i)
```

# Nested Loops
### Nested Loops ---> It is going to have a loop inside a loop

#### F string ---> will be helping us out with printing the numbers wth the variable value  present inside the colons so we can directly represent the variable value inside the colons messege itself the braces

```python
### Nested loop can hep us with iterating over a series of a number as with the proper combination of numbers
### Proper combination of iterating number
## I want to print a number from 0 to 5  but format is 0,1 ;0,2

for i in range(6):
    for j in range(4):
        print(i,j)
```

```python
for i in range(6):
    for j in range(4):
        print(f"i value is : {i} and j value is :{j}")
```

```python
## printing the sum of first N natural numbers where natural number ----> 1, 2, 3, 4....N
n_value=int(input("Enter the value for n :"))
sum=0
for i in range(n_value+1):
    sum+=i
    print(f"The sum of first {n_value} natural number is:{sum}")
```

```python
### print the prime number from 0 to 100
for num in range(2,101):
    for j in range(2,num):
        if num%j==0:
            break
        else:
            print(num)
```
