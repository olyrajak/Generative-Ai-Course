#  Range function range()

## range() function can be used for working as our iterable 
## rang() take an input parameter and the first parameter that the range function will be taken as the input will be the initial value that you want to print or you want to operate on and then you eill be passing the next input after , as the parameter as the final value and it can take third parameter as well and that will be the step size that is the difference between the itarating values .

## range(initial_vaue,final_value+1,step_size(difference))


```python
# Print the numbers from 1 to 1000
## range(initial_vaue,final_value+1,step_size(difference))
for i in range(1,11,2):
    print(i)
```

## In The range() there is never a necessary for mentionaing the step size and similarly if you want to print numbers or you want to operate on numbers again there is no neccessecity  for passing the initial value as well if and only if your initial value is 0 

```python
# number from 0 to 100
for item in range(101):
    print(item)
```

```python
# to print the number  0 to 100 a gap of 5

for i in range(0,100,5):
    print(i)
```

### BY using range() we can represent the number in a decrement manners as well or in a decreasing order as well

```python
# Print number 50 to 1 gap of 1 value in a decrement manner
for item in range(50,0,-1):
    print(item)
```

```python
# print a number 50 to 0 gap of 5
for item in range(50,-1,-5):
    print(item)
```

```python
# print a number 500 to 0 gap of 5
for item in range(500,-1,-50):
    print(item)
```
