```python
# Print the numbers from 1 to 1000
## range(initial_vaue,final_value+1,step_size(difference))
for i in range(1,11,2):
    print(i)
```

### BY using range() we can represent the number in a decrement manners as well or in a decreasing order as well

```python
# Print number 50 to 1 gap of 1 value in a decrement manner
for item in range(50,0,-1):
    print(item)
```

```python
# print a number 50 to 0 gap of 5
for item in range(50,-1,-5):
    print(item)
```

```python
# print a number 500 to 0 gap of 5
for item in range(500,-1,-50):
    print(item)
```

# While Loop

while loop is being used for iterating over a series of numbers when a condition is true and initialize a value as well while ierating 

```python
## by making the use of while loop we need to print the number 0 to 50
i=0
while i<50:
    print(i)
    i=i+1
```

```python
# I have to print the numbers from 0 to 50 as the even number
for i in range(51):
    if i%2==0:
        print(i)
```

```python
# print even number using a while loop 0 to 50
i=0
while i<=50:
    if i%2==0:
        print(i)
    else:
        pass
    i=i+1
```
