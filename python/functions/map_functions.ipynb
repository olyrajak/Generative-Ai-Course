# Map()

### Map function applies a given function operation to all items in an iterable or a list  and returns a map object (An Iterator)

###  Map function is partially useful for transforming he data or applying an operation on all items of the iterable itself

#### Map() is going to help us out with avoiding unneccessary lines of code and we can apply the operation on all values or all items in an iterable and this is the sign of a good programmer

### Syntax for map function is 
### Map()
### list(map(function,iterable))


```python
## Squaring all items of a list
## list---> [1,2,3,4,5]
def square_list(lst):
    squared_items = []
    for item in lst:
        squared_items.append(item**2)
    return squared_items

print(square_list([1, 2, 3, 4, 5]))
```

```python
## Similar problems statement with a map()
list(map(lambda x: x**2, [3, 5, 6, 1, 2, 7]))

```

```python
## map function with basic function in it
def square(x):
    return x**2

print(list(map(square, [3, 5, 6, 1, 2, 7])))
```

#### Whenever you are ging to pass the general function inside your map function as the parameter then you dont have to use parenthesis

### While Mapping the values of one iterable with another iterable we can actually apply the operation in between the values of the two iterables in a syncronus way 

### [1,2,3,4]<--->[6,7,8,9]---->addition operations[7,9,11,13]


```python
## Map function with multiple iterables 

numbers_1 = [1, 2, 3, 4, 5]
numbers_2 = [6, 7, 8, 9, 10]
map_result = map(lambda x, y: x + y, numbers_1, numbers_2)
print(list(map_result))  # Output: [7, 9, 11,
```

```python
### Map() to convert all string elements of a list to uppercase
words = ['hello', 'world', 'python', 'programming']
print(list(map(str.upper, words)))
```

```python
### Map() to convert all string elements of a list to lowercase
words = ['HELLO', 'WORLD', 'PYTHON', 'PROGRAMMING']
print(list(map(str.lower, words)))
```

```python
### Map() to convert all strings values items in an iterable into an integer
map_result = map(int, ['1', '2', '3', '4', '5'])
print(list(map_result))  # Output: [1, 2, 3, 4, 5]
```

```python
### Map() to convert all integer values items in an iterable into an string
map_result = map(str, [1, 2, 3, 4, 5])
print(list(map_result))  # Output: ['1', '2', '3', '
```

```python
### Retrive the values inside a dictionary using Map()
my_dict = [{'a': 1, 'b': 2, 'c': 3}, {'a': 4, 'b': 5, 'c': 6},
           {'a': 7, 'b': 8, 'c': 9}]
# Using map() to retrieve values from the dictionary
map_result = map(lambda d: d.get('b'), my_dict)
print(list(map_result))  # Output: [1, 4, 7]
```
