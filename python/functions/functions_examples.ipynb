```python
## To Create a function that convert temperature from  celcius to fahrenheit or Fahrenhite to celcius
## if temperature is in celcius then celcius convertion is : 
''' 
1 degree celceius = 9/5 + 32
 45 deg c = 45*9/5 +32

 1  farhenhiet celcius = 5/9 -32
45 deg F = (45-32 )*5/9 deg c
'''

def temparature_change(temp_value,unit):
    if unit== 'C':
        temperature=round(temp_value*9/5 + 32)
        print(f"The given temperature for {temp_value} deg celcius in Fahrenheit is :",str(temperature)  + " F")
    elif unit=='F':
        temperature=round((temp_value-32)*5/9)
        print(f"The given temperature for {temp_value} deg Fahrenheit in celcius  is :",str(temperature) + " C")
    else:
        print(f"Invalid Unit Enter by user side")
```

```python
temparature_change(1,'C')
```

```python
temparature_change(80,'F')
```
