```python
## To Create a function that convert temperature from  celcius to fahrenheit or Fahrenhite to celcius
## if temperature is in celcius then celcius convertion is : 
''' 
1 degree celceius = 9/5 + 32
 45 deg c = 45*9/5 +32

 1  farhenhiet celcius = 5/9 -32
45 deg F = (45-32 )*5/9 deg c
'''

def temparature_change(temp_value,unit):
    if unit== 'C':
        temperature=round(temp_value*9/5 + 32)
        print(f"The given temperature for {temp_value} deg celcius in Fahrenheit is :",str(temperature)  + " F")
    elif unit=='F':
        temperature=round((temp_value-32)*5/9)
        print(f"The given temperature for {temp_value} deg Fahrenheit in celcius  is :",str(temperature) + " C")
    else:
        print(f"Invalid Unit Enter by user side")
```

```python
temparature_change(1,'C')
```

```python
temparature_change(80,'F')
```

### isupper()----> function is check that wheather if the character is a upper case character or not

### isdigit()---> checks that wheather if the current character

```python
def password_strength(password):
    special_character='!@#%^&*()_'
    if len(password)<8:
        print(f"The entered password {password} is a weakest password because lenth is less than 8")

    elif not any (char.islower() for char in password ):
            print(f"The entered password {password} is a weakest password because there is not any lower case character")
    elif not any (char.isupper() for char in password ):
            print(f"The entered password {password} is a weakest password because there is not any Upper case character")
    elif not any (char.isdigit() for char in password ):
         print(f"The entered password {password} is a weakest password because there is not any digit")
         
    elif not any(char in special_character for char in password):
        print(f"The entered password {password} is a weakest password because there is not any special character")

    else:
        print(f"The entered password {password} is a strongest password")
 


```
