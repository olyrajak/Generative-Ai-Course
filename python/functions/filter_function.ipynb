### Filter function


### Filter functionis going to retieve only those items which are meeting a certain condition in an iterable

### The filter function constructs an iterator from elements of an iterable for which a condition remains true or for which a function returns true value.

#### Syntax for filter function

#### list(filter(func,iterable))

```python
### To Retrive only those values which are even 
def even_function(value):
    if value % 2 == 0:
        return True
    else:
        return False
```

```python
list_of_numbers = [1, 21, 3, 4, 5, 6, 7, 8, 9,44,66,12,3,4,28,2,34,55, 10]
filter_result = filter(even_function, list_of_numbers)
print(list(filter_result))  # Output: [4, 6, 8, 44, 66, 12, 4, 28, 2, 34, 10]
```

```python
### Solving the similar problem statement with a lamda function
print(list(filter(lambda x: x % 2 == 0, list_of_numbers)))  # Output: [4, 6, 8, 44, 66, 12, 4, 28, 2, 34, 10]
```

```python
### I want to retrive whch the value is > 5

print(list(filter(lambda x: x > 5, list_of_numbers)))  # Output: [21, 6, 7, 8, 9, 44, 66, 12, 28, 34, 55, 10]
```

```python
#### I want to retrive all those values which are greater than 5 as  well  even
list_numbers = [1, 21, 3, 4, 5, 6, 7, 8, 9,44,66,12,3,4,28,2,34,55, 10]
print(list(filter(lambda x: x > 5 and x % 2 == 0, list_numbers)))
```

```python
## Iterating over the dictionary items

employes = [
    {'name': 'Alice', 'age': 30, 'qualification': 'HR'},
    {'name': 'Bob', 'age': 25, 'qualification': 'IT'},
    {'name': 'Charlie', 'age': 26, 'qualification': 'Finance'},
    {'name': 'David', 'age': 28, 'qualification': 'IT'},
    {'name': 'Eve', 'age': 32, 'qualification': 'HR'}]


```

```python
def filter_employee_by_age(employee):
    if(employee['age'] > 25):
        return True
        
```

```python
filter_age=list(filter(filter_employee_by_age,employes))
print(filter_age)


```

```python
def filter_employees_by_qualification(employees, qualification):
    return list(filter(lambda employee: employee['qualification'] == qualification, employees))
```

```python
print(filter_employees_by_qualification(employes, 'IT'))  # Output: [{'name': 'Bob', 'age': 25, 'qualification': 'IT'}, {'name': 'David', 'age': 28, 'qualification': 'IT'}]
```
