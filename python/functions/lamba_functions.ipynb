### Lamda function are the 1 liner functions  defined by the lamda keyword 

## Syntax for lamda FUnction is :
## lamda arguments : expression 
## lamda ---> keyword which simply let the python interpreter know that the current function is our lamda function 
## Arguments ---> Input values for our features
## Expression ----> Logic ( Function Body )


#### lamda function example:
#### x:x**2


## Lamda function that can have any number of argument but only one expression or only one logic 
## They are used for short Operations or as Arguments to higher order Function

```python
## function that calculate square of numbers

def square(num):
    return num**2
```

```python
print(square(2))
```

```python
lambda_function=lambda x:x**2
```

```python
type(lambda_function)
print(lambda_function(44))
```

```python
# lamda function to calculate the sum of two numbers

sum=lambda x,y:x+y
print(sum(2,3))
```

```python
# product of two number
product= lambda x,y,z:x*y*z
print(product(2,3,4))
```
