```python
def translator(English_text):
    """This function is going to be used for converting the english language to hindi language"""

    if English_text=="Hello" or English_text=="hello":
        print(f"The translation for {English_text} is Namasthe")

    if English_text=="Help" or English_text=="help":
        print(f"The transalation for {English_text} is Mdad")

    if English_text=="Required" or English_text=="required":
        print(f"The transalation for {English_text} is Abaskayak")

```

### For having the access to our function we can simply and to provide the input to our function we can simply call the function

### Function calling is just like calling a waiter in a resturant where in just like we call the waitor and provide the input as the food item and then waitor provide us with the food itself , like in a similar way we can call a function as well and can provide the input and an get the required output after calling the function only

```python
## Calling a function --->Mention the name of the function and then provide the input

English_Input=input("Enter Your  English sentence for getting it transalated into hindi")

translator(English_Input)
```
