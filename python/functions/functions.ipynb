# Recursive Function

## By a Recursive Function we can say that a function i going to call itself again and again

```python
## Factorial Of a number
## Factorial of 8 ==8*7*6*5*4*3*2*1
## Factorial of 98 == 98*97*96....*1
## Factorial of 0 or 1 is 1
```

```python
def factorial(num):
    if num==0 or num==1:
        print(f"The factorial of {num} is 1")
        
    else:
        factorial=1
        for i in range(1,num+1):
            factorial=factorial*i
        print(f"The Factorial of {num} is :{factorial}")



               
```

```python
number=int(input("Enter Number"))
factorial(number)
```

```python
def factorial_recursive(num):
    if num==0 or num==1:
       return 1
    else:
        return num*factorial_recursive(num-1)
```

```python
factorial(5)
```

```python
## Reurn num* factorial_recusive(num-1)
''' 
base consition ----> which will stop the loop which will let the function to stop calling itself 
in our case , base condiiton was ---> if i==0 or i==1 ---> 1
num* factorial_recursive(num-1)
5*factorial_recussive(4) & factorial_recussive(4)--->4*factorial_recussive(3)& factorial_recussive(4)--->4*factorial_recussive(3)& factorial_recussive(3)--->3*factorial_recussive(2)& factorial_recussive(2)--->2*factorial_recussive(1)
'''
```

```python
input_ecussive_value=int(input("Enter the value : "))
print(f"The factorial of the given number {input_ecussive_value} : ",factorial_recursive(input_ecussive_value))
```

### Function with variables length Arguments
### Positional Argument ---> Position Arguments are being used so that we can pass multiple  values inside our function as an argument irrespective of the number of values we have mentioned as the parameter inside our function

```python
## Function to add multiple numbers irrespectve of how many number we are passing

def addition_positional(*num):
    add=0 
    for i in num:
        add+=i
    print(f"Addition of the numbers are :",add)


```

```python
addition_positional(3,4,5,6,7)
```

```python
addition_positional(1)
```

```python
addition_positional(444,4,44,44,667,888,8901,10)
```

### Keyword Arguments ----> Takes The input as the key words values and irrespective of any number of keyword values you are going to pass it wont matter

```python
def data(**user_data):
    for key,value in user_data.items():
        print(f"{key}:{value}")
        
```

```python
data(name="Oly",age=22,gender="Female")
```

```python
data(name="Oly",age=22,gender="Female",Qualification="MCA")
```

### You want to pass keyword value pairs as your input along with positional arguments as well 

```python
def variable(*positional,**keyword):
    for i in positional:
        print(f"Positional Values are :" ,i )
    for key,value in keyword.items():
        print(f"key is : {key} and value is : {value}")
```

```python
variable(1,2,3,4,5,name="Oly",age=22,gender="Female",Qualification="MCA")
```
