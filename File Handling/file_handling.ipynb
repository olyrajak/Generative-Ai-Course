```python
## open() ---> opening your file 
## First Parameter ---> File Name 
## Second Parameter ---> Mode of operation (r, w, a, r+, w+, a+)
'''
 Mode is going to decide the way in which we are working with this file
 Mode ---> r --> reading, w --> writing, a--> appending, r+ --> reading and writing, w+ --> writing and reading, a+ --> appending and reading
'''
```

```python
## Reading your file
## read() ---> reading the content of the file
## content = open_file.read()
```

```python
open_file=open('example.txt', 'r')
read_file=open_file.read()
print(read_file)
open_file.close()
```
