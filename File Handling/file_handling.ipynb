```python
## open() ---> opening your file 
## First Parameter ---> File Name 
## Second Parameter ---> Mode of operation (r, w, a, r+, w+, a+)
'''
 Mode is going to decide the way in which we are working with this file
 Mode ---> r --> reading, w --> writing, a--> appending, r+ --> reading and writing, w+ --> writing and reading, a+ --> appending and reading
'''
```

```python
## Reading your file
## read() ---> reading the content of the file
## content = open_file.read()
```

```python
open_file=open('example.txt', 'r')
read_file=open_file.read()
print(read_file)
open_file.close()
```

### with ---> It will be used for first of all opening the file as a object and there is no requirement for closing the file as well

```python
with open('example.txt', 'r') as open_file:
    read_file=open_file.read()
    print(read_file)
```

```python
with open('example2.txt', 'w') as file:
    read_file=file.write("WE are exploring the file handling in python")
    print(read_file)
```

```python
with open('example2.txt', 'a') as open_file:
    open_file.write('\nHello we are exploring file handling in python\n')  
```

```python
## Write the content into our file line by line ----> writelines() ---> writing the content of the file line by line

```

```python
lines=['First_line\n', 'Second_line\n', 'Third_line\n']
with open('example3.txt', 'w') as open_file:
    open_file.writelines(lines)
```

```python
### Mode ----> write the content into the binar file ----> wb
### Mode ----> Read the content from the binary file ----> rb
```

```python
data=b'\x00\x01\x02\x03\x04\x05\x06\x07\x08\t\n'
with open('example3.txt', 'wb') as binary_data:
    binary_data.write(data)
```

```python
with open('example3.txt', 'rb') as binary_data:
    read_data=binary_data.read()
    print(read_data)
```

```python
with open('source.txt','r') as source:
    content=source.read()

with open('destination.txt','w') as destination:
    destination.write(content)
```
