# OS LIBRARY

#### OS Library is able to interact with our operating system and can let us work with Multiple Directories and files

```python
## getcwd()---> It is capable of providing us with the idea about the current working directory  
### getcwd() ---> Current working directory
```

```python
import os
```

```python
current_path=os.getcwd()
print("Current Working Directory is:",current_path)
```

```python
## mkdir()--->  IT Stand for make directory. It is used to create a new directory in the current working directory.
```

```python
new_path="New Folder"
os.mkdir(new_path)
print("New Folder Created in Current Working Directory:",new_path)
```

```python
## listdir()---> it listing all of the directories present in the current working directory.
```

```python
print("List of Directories in Current Working Directory:",os.listdir('.'))
```

```python
## Joinning 2 or multiple  paths Together
os.path.join("C:\\Users\\Admin\\Desktop","New Folder")
```

```python
dir_name="Folder_data"
file_name="data.txt"
os.path.join(dir_name,file_name)
print("Joinning 2 or multiple  paths Together:",os.path.join(dir_name,file_name))
```

```python
### I have to join multiple path with the current path
current_path=os.getcwd()
dir_name="Folder_data"
file_name="data.txt"
os.path.join(current_path,dir_name,file_name)
os.mkdir(os.path.join(current_path,dir_name,file_name))
print("Joinning 2 or multiple  paths Together with Current Path:",os.path.join(current_path,dir_name,file_name))
```

```python
## os.path.exists()---> It is used to check whether the specified path exists or not. It returns True if the path exists, otherwise it returns False.
given_path="C:\Generative Ai Course\File Handling\OS_Library.ipynb"
os.path.exists(given_path)
if os.path.exists(given_path):
    print("The Given Path Exists:",given_path)
else:
    print("The Given Path Does Not Exists:",given_path)
```

```python
### os.path.isfile()---> It is used to check whether the specified path is a file or not. It returns True if the path is a file, otherwise it returns False.

### os.path.isdir()---> It is used to check whether the specified path is a directory or not. It returns True if the path is a directory, otherwise it returns False.
```

```python
path="total_data.txt"
if os.path.isfile(path):
    print("The Given Path is a File:")
else:
    print("The Given Path is Not a File:")
```

```python
paths="New Folder"
if os.path.isdir(paths):
    print("The Given Path is a Directory:")
else:
    print("The Given Path is Not a Directory:")
```

```python
### The complete path is going to be <---> Absolute path ---> os.path.abspath()

# os.path.abspath()---> It Provide us with the complete path of the absolute path of the file or directory
```

```python
file_name="total_data.txt" #----> Relative path
complete_path=os.path.abspath(file_name)
print(complete_path)

```
